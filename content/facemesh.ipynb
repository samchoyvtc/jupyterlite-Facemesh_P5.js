# p5.js FaceMesh (45 minutes)

The sketch draws points on your face. The camera picture stays in this browser.

1. Run each code cell from top to bottom (Shift+Enter).
2. When the browser asks, allow the camera.
3. Wait until green points sit on your face. The first load downloads the model.

| Time | What you do |
| --- | --- |
| 0–8 min | Run the cells below until the points appear |
| 8–18 min | Read the four ideas, and match them to the cells |
| 18–30 min | Change the point size and color, then run that cell again |

## Four ideas

- `setup` runs once. It makes the canvas and starts the camera.
- `draw` runs over and over. It paints the camera picture and the points.
- `ml5.faceMesh` finds one face.
- A loop draws one circle for each face point.


## Edit these numbers

Run this cell again after you change it. The sketch updates.

- **18–30 min.** Change `pointSize`, `pointR`, `pointG`, and `pointB`.
- **30–42 min.** Set `showNose` and `showEyes` to `true`.

Nose tip is point **1**. The eyes are points **33** and **263**.


```p5js
// 18-30 min: size and color of the small face points
var pointSize = 4;
var pointR = 0;
var pointG = 255;
var pointB = 80;

// 30-42 min: set either of these to true
var showNose = false;
var showEyes = false;
```

## The picture

`setup` starts the camera. `draw` paints it, then the points. Nose and eye circles appear only after you turn them on.


```p5js
var faceMesh = null;
var video = null;
var faces = [];
var modelReady = false;

function setup() {
  createCanvas(640, 480);
  video = createCapture(VIDEO);
  video.size(640, 480);
  video.hide();
  loadMl5();
}

function draw() {
  background(20);
  if (video) {
    push();
    translate(width, 0);
    scale(-1, 1);
    image(video, 0, 0, width, height);
    pop();
  }
  if (!modelReady) {
    fill(255);
    noStroke();
    textSize(18);
    text("Loading face model...", 20, 40);
    return;
  }
  if (faces.length === 0) {
    fill(255);
    noStroke();
    textSize(18);
    text("No face yet. Look at the camera.", 20, 40);
    return;
  }
  drawPoints();
  if (showNose) {
    drawNose();
  }
  if (showEyes) {
    drawEyes();
  }
}

function drawPoints() {
  var i;
  var j;
  var face;
  var kp;
  var pos;
  fill(pointR, pointG, pointB);
  noStroke();
  for (i = 0; i < faces.length; i = i + 1) {
    face = faces[i];
    for (j = 0; j < face.keypoints.length; j = j + 1) {
      kp = face.keypoints[j];
      pos = place(kp);
      circle(pos.x, pos.y, pointSize);
    }
  }
}

function drawNose() {
  var pos = place(faces[0].keypoints[1]);
  fill(255, 80, 80);
  noStroke();
  circle(pos.x, pos.y, 22);
}

function drawEyes() {
  var leftPos = place(faces[0].keypoints[33]);
  var rightPos = place(faces[0].keypoints[263]);
  fill(80, 160, 255);
  noStroke();
  circle(leftPos.x, leftPos.y, 18);
  circle(rightPos.x, rightPos.y, 18);
}

function place(kp) {
  var sx = 1;
  var sy = 1;
  if (video && video.width > 0 && video.height > 0) {
    sx = width / video.width;
    sy = height / video.height;
  }
  return { x: kp.x * sx, y: kp.y * sy };
}
```

## The face model

This cell loads ml5, then asks FaceMesh to watch the camera. When a face is found, `faces` is updated and `draw` paints it.


```p5js
var ML5_URL = "https://cdn.jsdelivr.net/npm/ml5@1.2.1/dist/ml5.min.js";

function loadMl5() {
  if (window.ml5) {
    startFaceMesh();
    return;
  }
  var script = document.createElement("script");
  script.src = ML5_URL;
  script.onload = startFaceMesh;
  script.onerror = function () {
    console.error("Could not load ml5");
  };
  document.head.appendChild(script);
}

function startFaceMesh() {
  faceMesh = ml5.faceMesh({ maxFaces: 1, flipped: true }, function (model) {
    faceMesh = model;
    faceMesh.detectStart(video, function (results) {
      faces = results;
    });
    modelReady = true;
  });
}
```
